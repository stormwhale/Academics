Re-creating the Krackhardt kite graph

```python
import networkx as nx
```

```python
#Define the graph:
G = nx.Graph()

#Create edges. Creating nodes are not needed unless an isolated node is desired.
edge_list = [
    ('Andre', 'Carol'), ('Andre', 'Beverly'), ('Andre', 'Diane'), ('Andre', 'Fernando'),\
    ('Carol', 'Fernando'), ('Carol', 'Diane'),\
    ('Fernando', 'Diane'), ('Fernando', 'Garth'), ('Fernando', 'Heather'),\
    ('Heather', 'Garth'), ('Heather', 'Ike'),\
    ('Ike', 'Jane'),\
    ('Garth', 'Diane'), ('Garth', 'Ed'), ('Garth', 'Beverly'),\
    ('Ed', 'Diane'), ('Ed', 'Beverly'),\
    ('Beverly', 'Diane'), ('Beverly', 'Garth')]
G.add_edges_from(edge_list)

#Force the positions of the nodes by using x and y axis coordinates.
pos = {
    'Andre':    (1, 3), 'Beverly': (1, 1),
    'Carol':    (2, 4), 'Diane':   (2, 2), 'Ed': (2, 0),
    'Fernando': (3, 3), 'Garth':   (3, 1),
    'Heather':  (4, 2),
    'Ike':      (5, 2),
    'Jane':     (6, 2)
}

nx.draw(G, pos, with_labels=True, node_size= 1000)
```
